```python
import h5py
import matplotlib as plt
import torch 
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn
import torch.nn.functional as f
```

```python
class BatteryDataset(Dataset):
    def __init__(self, hdf5_path, transform=None):
        self.hdf5_path = hdf5_path
        self.transform = transform

    def __getitem__(self, idx):
       
        with h5py.File(self.hdf5_path, 'r') as f:
            image = f['image'][idx]
            label_str = f['battery_typ'][idx].decode('utf-8')
            
        if self.transform:
            image = self.transform(image)

        label_map = {'cylindrical': 0, 'prismatic': 1, 'pouch': 2}
        label_int = label_map[label_str]
            
        # Converts into Pytorch tensors
        return torch.tensor(image, dtype=torch.float32), torch.tensor(label_int, dtype=torch.long)
    
    def data_exploration(self):
        with h5py.File(self.hdf5_path, 'r') as f:
            print(f'total ammount of images: {len(f['image'])}')
            
            labels = f['battery_typ']
            print(f'all the possible labels: {labels}')
```

```python
dataset1 = BatteryDataset('data.h5')
```
