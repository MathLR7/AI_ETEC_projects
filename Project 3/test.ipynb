```python
class BatteryDataset(Dataset):
    def __init__(self, hdf5_path, transform=None):
        self.hdf5_path = hdf5_path
        self.transform = transform
        
        # Inspeciona o tamanho do dataset
        with h5py.File(self.hdf5_path, 'r') as f:
            # Substitua 'images' e 'labels' pelas chaves exatas do seu dataset
            self.length = len(f['labels']) 
            
    def __len__(self):
        return self.length

    def __getitem__(self, idx):
        # Abrimos o arquivo dentro do __getitem__ para funcionar bem com num_workers > 0 no DataLoader
        with h5py.File(self.hdf5_path, 'r') as f:
            image = f['images'][idx]
            label = f['labels'][idx]
            
        if self.transform:
            image = self.transform(image)
            
        # Converte para tensores do PyTorch
        return torch.tensor(image, dtype=torch.float32), torch.tensor(label, dtype=torch.long)
```
